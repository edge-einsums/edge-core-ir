The `(max, min)` mirror of `(min, +)` Bellman-Ford: for each vertex it finds the path from the source that **maximizes the minimum edge capacity** along it.

```
F[i+1, d] = G[s,d] . F[i,s]   :: map(min, intersect) reduce(max, union)
D[i+1, d] = D[i,d] . F[i+1,d] :: map(max, union) reduce(max, union)
stop: stable(D)
```
